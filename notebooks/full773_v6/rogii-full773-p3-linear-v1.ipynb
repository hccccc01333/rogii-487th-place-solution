```python
import hashlib
SRC = "/kaggle/input/full773-submission-candidates-lf/p3_linear_lf.csv"
EXPECT = "a59693cd87c16e9261b270986ce27656122f87417f8f7befde698d8b68e2f21c"
data = open(SRC, "rb").read()
assert hashlib.sha256(data).hexdigest() == EXPECT, "sha mismatch"
print("input ok", len(data), hashlib.sha256(data).hexdigest())
```

```python
import glob
import hashlib
import os
import numpy as np
import pandas as pd

sub = pd.read_csv(SRC)
print("input_dir:", os.listdir("/kaggle/input") if os.path.isdir("/kaggle/input") else "MISSING")
cands = sorted(glob.glob("/kaggle/input/**/sample_submission.csv", recursive=True))
print("sample candidates:", cands)
if cands:
    sample = pd.read_csv(cands[0])[["id"]]
    out = sample.merge(sub, on="id", how="left")
else:
    out = sub[["id", "tvt"]].copy()
assert list(out.columns) == ["id", "tvt"], list(out.columns)
assert out["id"].is_unique
assert out["tvt"].notna().all()
assert np.isfinite(out["tvt"].to_numpy(dtype=float)).all()
out[["id", "tvt"]].to_csv("/kaggle/working/submission.csv", index=False)
raw = open("/kaggle/working/submission.csv", "rb").read()
print("wrote", len(raw), hashlib.sha256(raw).hexdigest())
print("rows", len(out), "pandas", pd.__version__)
```
